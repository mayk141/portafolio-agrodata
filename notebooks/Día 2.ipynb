```python
# Día 2: Listas y Diccionarios
# LISTAS: orden + índice

ph_lotes = [6.2, 5.8, 7.1, 6.5]

print("Primer elemento (índice 0):")
print(ph_lotes[0])

print("\nÚltimo elemento (índice -1):")
print(ph_lotes[-1])

print("\nAgregar un valor nuevo:")
ph_lotes.append(6.9)
print(ph_lotes)
```

```python
# DICCIONARIOS: clave-valor, sin orden garantizado

lote = {"nombre": "Lote_3", "cultivo": "Café", "area_ha": 4.2}

print("Acceder por clave:")
print(lote["cultivo"])

print("\nAgregar una clave nueva:")
lote["ph"] = 6.1
print(lote)

print("\nVer todas las claves:")
print(lote.keys())
```

```python
finca = {
    "nombre": "Finca Las Brisas",
    "lotes": [
        {"id": "L1", "cultivo": "Café", "area_ha": 3.5},
        {"id": "L2", "cultivo": "Maíz", "area_ha": 5.1},
        {"id": "L3", "cultivo": "Café", "area_ha": 2.8},
    ]
}

area_total = 0
cultivos_presentes = []

for lote in finca["lotes"]:
    area_total += lote["area_ha"]
    if lote["cultivo"] not in cultivos_presentes:
        cultivos_presentes.append(lote["cultivo"])

print(f"Finca: {finca['nombre']}")
print(f"Área total: {round(area_total, 2)} ha")
print(f"Cultivos presentes: {cultivos_presentes}")
```
